### 항공기 지연데이터

```python
import pandas as pd

url = "https://raw.githubusercontent.com/datadigger01/AppStatMat/main/Data/FlightDelays.csv"
df = pd.read_csv(url)
display(df.head())
```

```python
print(df.info())
df
```

## 1. 컬럼명 타입 변환
- FL_DATE 컬럼은 date 타입으로 변환

```python
df['FL_DATE'] = pd.to_datetime(df['FL_DATE'])

display(df.info())
df
```

## 2. Target 컬럼 생성
- 컬럼명 'Fight Status'의 값이 'ontime'이면 숫자 0 나머지(지연) 1으로 생성하여 delay_yn 컬럼명으로 새로 생성

```python
df['delay_yn'] = (df['Flight Status'] != 'ontime').astype(int)
display(df.head())
```

```python
df.columns
```

## 3. 결측값 확인 및 대체

```python
missing_values_count = df.isnull().sum()
display(missing_values_count)
```

## 4.새로운 변수 파생
- CRS_DEP_TIME을 100으로 나누어 반올림(round) 처리 : 출발예정시간으로 오전 6:00 부터 10:00 사이를 18개의 그룹 등간격

- Fl_DATE 컬럼으로부터 월 추출

```python
df['CRS_DEP_TIME_G'] = round(df['CRS_DEP_TIME'] / 100).astype(int)

#df['DEP_MONTH'] = df['FL_DATE'].dt.month
print(df.info())
display(df)
```

```python
# 그룹별 발생빈도 확인
df['CRS_DEP_TIME_G'].value_counts()
```

## 5.데이터탐색(EDA)
   - 범주형 데이터(Categorical)

```python
# prompt: 'CRS_DEP_TIME', 'CARRIER', 'DEST', 'ORIGIN', 'Weather', 'DAY_WEEK', 'DAY_OF_MONTH' 컬럼에 대해서 왼쪽축 기준으로는 정렬된 컬럼값 기준의 빈도수를
# 오른쪽축기준으로는 'delay_yn' 값에 대한 발생비율이 최대값이 1.0으로 고정된 그래프로 그리되, 3열로 보여지게 하는 코드를 제공해주세요.
# 그리고 막대그래프를  연한회색으로하고 선그래프는 파란색으로 셋팅해주세요

import matplotlib.pyplot as plt

# 컬럼 리스트
columns_to_plot = ['CRS_DEP_TIME_G','CARRIER', 'DEST', 'ORIGIN', 'Weather', 'DAY_WEEK', 'DAY_OF_MONTH']

# 3열로 그래프 그리기 위한 설정
num_cols = 2
num_rows = (len(columns_to_plot) + num_cols - 1) // num_cols

# 그래프 그리기
fig, axes = plt.subplots(num_rows, num_cols, figsize=(14,  5 * num_rows))
fig.tight_layout(pad=5.0)

for i, column in enumerate(columns_to_plot):
  row = i // num_cols
  col = i % num_cols

  # 범주형 변수에 대해 빈도수 및 Tgt_3YN 비율 계산
  value_counts = df[column].value_counts().sort_index()
  tgt_3yn_ratio = df.groupby(column)['delay_yn'].mean()

  # 왼쪽 축에 막대 그래프 그리기 (빈도수)
  axes[row, col].bar(value_counts.index, value_counts.values, color='lightgray')
  axes[row, col].set_xlabel(column)
  axes[row, col].set_ylabel('Frequency')

  # 오른쪽 축에 선 그래프 그리기 (delay_yn 비율)
  axes2 = axes[row, col].twinx()
  axes2.plot(tgt_3yn_ratio.index, tgt_3yn_ratio.values, color='darkred')
  axes2.set_ylabel('delay_yn Ratio')
  axes2.set_ylim(0, 1.0)  # y축 범위를 0에서 1.0 으로 고정

  axes[row, col].set_title(f'{column} Frequency and delay_yn Ratio')
  axes[row, col].tick_params(axis='x', labelsize=8)

  # 빈 공간 제거
  if i >= len(columns_to_plot):
    fig.delaxes(axes[row, col])


plt.show()
```

- 연속형(Continous) 데이터를 범주화 시켜서

```python
# Categorize the Distance column into 10 groups
df['Distance_Group'] = pd.qcut(df['DISTANCE'], q=10, labels=False, duplicates='drop')

# Group by the new Distance_Group and calculate frequency and delay ratio
value_counts_distance = df['Distance_Group'].value_counts().sort_index()
delay_ratio_distance = df.groupby('Distance_Group')['delay_yn'].mean()

# Create the plot
fig, ax1 = plt.subplots(figsize=(10, 6))

# Bar plot for frequency on the left axis
ax1.bar(value_counts_distance.index, value_counts_distance.values, color='lightgray')
ax1.set_xlabel('Distance Group')
ax1.set_ylabel('Frequency')
ax1.set_title('Distance Group Frequency and delay_yn Ratio')
ax1.tick_params(axis='x', rotation=45)

# Create custom labels for the x-axis ticks
distance_bins = df.groupby('Distance_Group')['DISTANCE'].agg(['min', 'max'])
xtick_labels = [f'{row["min"]}-{row["max"]}' for index, row in distance_bins.iterrows()]
ax1.set_xticks(value_counts_distance.index)
ax1.set_xticklabels(xtick_labels, rotation=45, ha='right')


# Line plot for delay_yn ratio on the right axis
ax2 = ax1.twinx()
ax2.plot(delay_ratio_distance.index, delay_ratio_distance.values, color='darkred')
ax2.set_ylabel('delay_yn Ratio')
ax2.set_ylim(0, 1.0)

plt.show()
```
